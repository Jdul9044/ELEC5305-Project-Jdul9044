# ELEC3505 - How MVDR beamforming performance can be improved compared to conventional beamforming for underwater acoustic signal

## 1. Load Library

```python
import matplotlib.pyplot as plt
import numpy             as np
import pandas            as pd
from matplotlib.gridspec import GridSpec
from scipy.signal        import stft, istft, spectrogram
from pathlib             import Path
from scipy.signal        import welch, decimate
from scipy.io            import wavfile, loadmat
```

## 2. Setup

### Input Parameters

```python
azi_deg = 171 #deg (main target)
# azi_deg = 294 #deg (inteference)
# azi_deg = 309 #deg (other)
ele_deg = 0 #deg
c = 1500 #m/s
nfft = 1028
hop = round(nfft/2)
window_len = 80 # window size for creating covariance matrix (MVDR beamforming)
simulation_mode = False # if user decides to test simulation data or use real underwater acoustic data
# simulation_mode = True # if user decides to test simulation data or use real underwater acoustic data
```

### Load Data (Simulation/Real Underwater Recording)

```python
################################################################################################
if simulation_mode == True:
    fs = 48e3 # to match sampling frequency 
    
    folder_path = r"Simulation_Data\Moving" #load folder location
    folder = Path(folder_path)
    
    signal_list = [] #create list
    
    for file in folder.glob("*.pcm"):
        print(file)
        signal = np.fromfile(file, dtype=np.int16) # load each 16-bit PCM
        
        # downsample
        signal = decimate(signal, int(fs/2000)) # downsample to 2000 Hz
        signal_list.append(signal)
        
    fs = 2000
    signal_array = np.array(signal_list) # convert into array
    
    mic_num = signal_array.shape[0]
    spacing = 1.875 #m
    mic_pos = np.zeros((mic_num,3))
    mic_pos[:, 1] = np.arange(mic_num) * spacing
###############################################################################################
else:
    fs = 3276.8 #HLA
    # fs = 1500 #VLA
    
    # file_path = "Real_Data\VLA_100000_SAMPLES.mat"
    file_path = "Real_Data\HLA_100000_SAMPLES.mat"
    signal_array = loadmat(file_path)['ans'].T
    # signal_array = signal_array[::-1,:] # FOR VLA ONLY
    
    # mic_pos = pd.read_excel("Real_Data\Hyd_Pos_VLA.xlsx",header=None).to_numpy()
    mic_pos = pd.read_excel("Real_Data\Hyd_Pos_HLA.xlsx",header=None).to_numpy()
    mic_pos = mic_pos[:,1:4]
    
    # select hydrophones
    signal_array = signal_array[0:5,:]
    mic_pos = mic_pos[0:5,:]

    mic_num = signal_array.shape[0]
    print(mic_num)
################################################################################################
```

### Normalise data for Full Scale

```python
signal_array = signal_array / np.max(np.abs(signal_array)) # normalise to full scale
```

## 3. Data Check

### Check if Data Clipping

```python
for i in range(signal_array.shape[0]):
    plt.plot(signal_array[i,:])
    plt.title("Time Domain of each Hydrophone")
    plt.xlabel("Samples")
    plt.ylabel("Amplitude")
```

## 4. Compute Spectrogram

### Choose Window

```python
# selected_window=("kaiser", 2)
selected_window='hann' # hanning window
```

### Compute for each Hydrophone

```python
# Compute Spectrogram for all channels
spectrogram_list = []
    
for i in range(mic_num):

    # compute spectrogram with selected parameters
    freq_bins, time_bins, X = stft(
    signal_array[i, :],
    fs=fs,
    window=selected_window,
    nperseg=nfft,
    noverlap=nfft - hop,
    nfft=nfft,
    return_onesided=True,
    boundary=None,
    padded=False,
    scaling='psd'
)
    
    spectrogram_list.append(X)
    
spectrogram_array = np.array(spectrogram_list)
del spectrogram_list # save memory

spectrogram_psd = np.abs(spectrogram_array)**2
spectrogram_psd[:,1:-1,:] *= 2 # since half energy is missing because negative frequency STFT coefficients are missing
```

### Check Channel Spectrogram

```python
selected_mic = 2 # select channel
```

```python
# Check Single Channel
plt.imshow(
    10*np.log10(spectrogram_psd[selected_mic,:,:]),
    aspect='auto',
    origin='lower',
    vmin = np.percentile(10*np.log10(spectrogram_psd[selected_mic,:,:]),5),
    vmax = np.percentile(10*np.log10(spectrogram_psd[selected_mic,:,:]),95),
    extent=[time_bins[0], time_bins[-1], freq_bins[0], freq_bins[-1]]
)

plt.title("Channel " + str(selected_mic) + " Hydrophone")
plt.xlabel("Elapsed Time (s)")
plt.ylabel("Frequency (Hz)")
plt.colorbar()
plt.ylim(0,500)
plt.xlim(0,10)
plt.ylim(100,300)
plt.show()
```

### Check Welch

```python
#Check Graph - for single microphone/hydrophone
f, Pxx = welch(signal_array[selected_mic,:], fs=fs, window=selected_window,
               nperseg=nfft, noverlap=nfft-hop, nfft=nfft, detrend=False)

plt.plot(f, 10*np.log10(Pxx))
plt.xlim(0,500)
# plt.ylim(-20,20)
plt.title("Channel " + str(selected_mic) + " Hydrophone, NFFT=" + str(nfft) + ", Overlap=50%")
plt.ylabel('PSD (dBFS/Hz)')
plt.xlabel("Frequency (Hz)")
plt.show()
```

## 5. Azimuth Scan

```python
azi_scan_deg = np.arange(0, 360, 1) # scan all 360 degrees of rotation
ele_scan_deg = np.zeros_like(azi_scan_deg) # set elevation to 0 degrees
azi_scan = np.deg2rad(azi_scan_deg) #rad
ele_scan = np.deg2rad(ele_scan_deg) #rad
```

```python
selected_freq_bin = 112 # select a frequency to do azimuth scan
freq_bin_idx = np.argmin(np.abs(freq_bins - selected_freq_bin))
```

```python
Y_freq = np.zeros((len(time_bins),len(azi_scan)), dtype=complex) # create array
Y_freq_mvdr = np.zeros((len(time_bins),len(azi_scan)), dtype=complex) # create array
for aa in range(len(azi_scan)):

        # unit vector for adjusted spherical coordinate system
        u = np.array([
            np.cos(ele_scan[aa]) * np.sin(azi_scan[aa]),
            np.cos(ele_scan[aa]) * np.cos(azi_scan[aa]),
            np.sin(ele_scan[aa])
            ])

        # time delay
        tau = -np.dot(mic_pos,u) / c

        # steering vector
        steering_vect = np.exp(-1j * 2 * np.pi* freq_bins[freq_bin_idx] * tau)
        # compute CBF output
        Y_freq[:,aa] = np.dot(np.conj(steering_vect), spectrogram_array[:,freq_bin_idx,:]) / mic_num

        # select a particular frequency bin with snapshots
        X_f = spectrogram_array[:, freq_bin_idx, :]
        # create covariance matrix
        R = (X_f @ X_f.conj().T) / X_f.shape[1]

        # apply adaptive loading
        loading = 0.1 * np.trace(R).real / mic_num
        # loading = 1e-3 * np.trace(R).real / mic_num
        R_loaded = R + loading * np.eye(mic_num)
        Rinv = np.linalg.inv(R_loaded)
        
        w = (Rinv @ steering_vect)/(steering_vect.conj().T @ Rinv @ steering_vect) # create MVDR weights                
        Y_freq_mvdr[:,aa] = np.dot(np.conj(w).flatten(), spectrogram_array[:, freq_bin_idx, :]) # compute MVDR output
```

```python
Y_freq = np.abs(Y_freq)**2 # convert into power
spectrogram_avg_azi = np.mean(Y_freq, axis=0) # time average
spectrogram_avg_azi_dB = 10*np.log10(spectrogram_avg_azi + 1e-12) # log scale

Y_freq_mvdr = np.abs(Y_freq_mvdr)**2 # convert into power
spectrogram_avg_azi = np.mean(Y_freq_mvdr, axis=0) # tine average
spectrogram_avg_azi_mvdr_dB = 10*np.log10(spectrogram_avg_azi + 1e-12) # log scale
```

```python
plt.figure()
plt.plot(azi_scan_deg,spectrogram_avg_azi_dB,label="CBF")
plt.plot(azi_scan_deg,spectrogram_avg_azi_mvdr_dB,label="MVDR")
plt.title(f"Azimuth Scan for Frequency {selected_freq_bin} Hz")
plt.xlabel("Azimuth (degrees)")
plt.ylabel("Magnitude (dB)")
plt.legend()
```

```python
peak_idx = np.argmax(spectrogram_avg_azi_mvdr_dB)
peak_angle = azi_scan_deg[peak_idx]
print(peak_angle) # find peak
```

## 6. Main Beamforming Setup

```python
azi = np.deg2rad(azi_deg) #rad
ele = np.deg2rad(ele_deg) #rad
```

### Initialise Array

```python
Y = np.zeros((len(freq_bins),len(time_bins)), dtype=np.complex128)
Y_mvdr = np.zeros((len(freq_bins),len(time_bins)), dtype=np.complex128)
```

### Calculate Time Delay (Tau)

```python
# Unit vector
u = np.array([
    np.cos(ele) * np.sin(azi),
    np.cos(ele) * np.cos(azi),
    np.sin(ele)
    ])
```

```python
# Time Delay
tau = -np.dot(mic_pos,u) / c
```

## 7. Perform Conventional Beamforming

```python
for tt in range(len(time_bins)):
    for ff in range(len(freq_bins)):
        steering_vect = np.exp(-1j * 2 * np.pi* freq_bins[ff] * tau)
        Y[ff,tt] = np.dot(np.conj(steering_vect), spectrogram_array[:,ff,tt]) / mic_num # normalise the amplitude after being added up
```

```python
Y_psd = np.abs(Y)**2 # convert into power
Y_psd[1:-1] *= 2 # account for half energy lost
Y_dB = 10*np.log10(Y_psd + 1e-12) # log scale
```

## 8. Perform MVDR Beamforming

```python
for ff in range(len(freq_bins)):
    # steering vector
    steering_vect = np.exp(-1j * 2 * np.pi * freq_bins[ff] * tau)
    steering_vect = steering_vect.reshape(-1, 1)

    for tt in range(len(time_bins)):
        # create movign window with across localised snapshots
        start_idx = max(0, tt - window_len)
        end_idx = min(len(time_bins), tt + window_len)

        # select frequency bin for covariance matrix
        X_f = spectrogram_array[:, ff, start_idx:end_idx]
        # compute covariance matrix
        R = (X_f @ X_f.conj().T) / X_f.shape[1]

        # adpative loading
        loading = 0.1 * np.trace(R).real / mic_num
        R_loaded = R + loading * np.eye(mic_num)
        Rinv = np.linalg.inv(R_loaded)
        
        w = (Rinv @ steering_vect)/(steering_vect.conj().T @ Rinv @ steering_vect) # compute MVDR weights
        Y_mvdr[ff, tt] = np.dot(np.conj(w).flatten(), spectrogram_array[:, ff, tt]) # compute MVDR output
```

```python
Y_mvdr_psd = np.abs(Y_mvdr)**2 # convert into power
Y_mvdr_psd[1:-1] *= 2 # account for half energy lost
Y_mvdr_dB = 10*np.log10(Y_mvdr_psd + 1e-12) # log scale
```

## 8. Compute Welch for Beamformer Outputs

```python
#Check Graph - Beamformer & MVDR Output
spectrogram_avg_bf = np.mean(np.abs(Y[:,:])**2, axis=1)
spectrogram_avg_bf[1:-1] *= 2 

spectrogram_mvdr_avg_bf = np.mean(np.abs(Y_mvdr[:,:])**2, axis=1)
spectrogram_mvdr_avg_bf[1:-1] *= 2 
```

## 9. Plot Welch Comparison

```python
plt.plot(f, 10*np.log10(Pxx), label="Channel " + str(selected_mic) + " Hydrophone")
plt.plot(freq_bins, 10*np.log10(spectrogram_avg_bf + 1e-12), label="Beamformer Output")
plt.plot(freq_bins, 10*np.log10(spectrogram_mvdr_avg_bf + 1e-12), label="MVDR Beamformer Output")
plt.xlim(0,400)
# plt.ylim(-20,20)
plt.title(f"Beam Focused at a={azi_deg} and e={ele_deg}, NFFT=" + str(nfft) + ", Overlap=50%")
plt.ylabel('Time Averaged PSD (dBFS/Hz)')
plt.xlabel("Frequency (Hz)")
plt.legend()
plt.show()
```

## 11. Compute WNG

```python
# to match ideal MVDR case (superdirective beamformer) the difference in the microphone positions is required
d = np.linalg.norm(mic_pos[:,None,:] - mic_pos[None,:,:], axis=2)
```

```python
azi_scan_deg = np.arange(0, 360, 1)
ele_scan_deg = np.zeros_like(azi_scan_deg)
azi = np.deg2rad(azi_scan_deg) #rad
ele = np.deg2rad(ele_scan_deg) #rad
```

```python
azi_target_deg = azi_deg
ele_target_deg = 0
azi_target = np.deg2rad(azi_target_deg) #rad
ele_target = np.deg2rad(ele_target_deg) #rad
```

```python
WNG = np.zeros(len(freq_bins)) # create array
WNG_mvdr = np.zeros(len(freq_bins)) #  create array
```

```python
# this will be the unit vector towards the target
u_target = np.array([
    np.cos(ele_target) * np.sin(azi_target),
    np.cos(ele_target) * np.cos(azi_target),
    np.sin(ele_target)
    ])

tau_target = -np.dot(mic_pos,u_target) / c
```

```python
for ff in range(len(freq_bins)):
    
    steering_vect_target = np.exp(-1j * 2 * np.pi* freq_bins[ff] * tau_target)
    # use weights to compute WNG
    WNG[ff] = 1 / np.real((steering_vect_target / mic_num).conj().T @ (steering_vect_target / mic_num)).item()

    # for idealised MVDR (superdirective beamformer), R is the diffuse coherence matrix
    R = np.sinc(2 * freq_bins[ff] * d / c) # sinc can be used to simplify the main equation
    loading = 0.1 * np.trace(R).real / mic_num # adaptive loading
    R_loaded = R + loading * np.eye(mic_num)
    Rinv = np.linalg.inv(R_loaded)
    w = (Rinv @ steering_vect_target)/(steering_vect_target.conj().T @ Rinv @ steering_vect_target)     
    # use MVDR weights to compute WNG
    WNG_mvdr[ff] = 1 / np.real(w.conj().T @ w).item()
```

```python
WNG = np.abs(WNG)
WNG_dB = 10*np.log10(WNG + 1e-12) # convert into log scale

WNG_mvdr = np.abs(WNG_mvdr)
WNG_mvdr_dB = 10*np.log10(WNG_mvdr + 1e-12) #convert into log scale
```

```python
plt.figure()
plt.plot(freq_bins,WNG_dB,label="CBF")
plt.plot(freq_bins,WNG_mvdr_dB,label="MVDR")
plt.title("White Noise Gain Comparison (Main Source)")
plt.xlim(0,400)
plt.xlabel("Frequency (Hz)")
plt.ylabel("Magnitude (dB)")
plt.legend()
```

## 12. Compute DI

```python
azi_scan_deg = np.arange(0, 360, 1)
ele_scan_deg = np.zeros_like(azi_scan_deg)
azi = np.deg2rad(azi_scan_deg) #rad
ele = np.deg2rad(ele_scan_deg) #rad
```

```python
azi_target_deg = azi_deg
ele_target_deg = 0
azi_target = np.deg2rad(azi_target_deg) #rad
ele_target = np.deg2rad(ele_target_deg) #rad
```

```python
Y_freq = np.zeros((len(freq_bins),len(azi)), dtype=complex)
Y_freq_mvdr = np.zeros((len(freq_bins),len(azi)), dtype=complex)

Y_target = np.zeros(len(freq_bins), dtype=complex)
Y_target_mvdr = np.zeros(len(freq_bins), dtype=complex)
```

```python
u_target = np.array([
    np.cos(ele_target) * np.sin(azi_target),
    np.cos(ele_target) * np.cos(azi_target),
    np.sin(ele_target)
    ])

tau_target = -np.dot(mic_pos,u_target) / c
```

```python
for ff in range(len(freq_bins)):

    # main target
    steering_vect_target = np.exp(-1j * 2 * np.pi* freq_bins[ff] * tau_target)

    R = np.sinc(2 * freq_bins[ff] * d / c)
    loading = 0.1 * np.trace(R).real / mic_num
    R_loaded = R + loading * np.eye(mic_num)
    Rinv = np.linalg.inv(R_loaded)
    w = (Rinv @ steering_vect_target)/(steering_vect_target.conj().T @ Rinv @ steering_vect_target)     

    Y_target[ff] = np.dot(np.conj(steering_vect_target), steering_vect_target) / mic_num
    Y_target_mvdr[ff] = np.dot(np.conj(w), steering_vect_target)

    for aa in range(len(azi)):

        # azimuth scan
        u = np.array([
            np.cos(ele[aa]) * np.sin(azi[aa]),
            np.cos(ele[aa]) * np.cos(azi[aa]),
            np.sin(ele[aa])
            ])
    
        tau = -np.dot(mic_pos,u) / c
    
        steering_vect = np.exp(-1j * 2 * np.pi* freq_bins[ff] * tau)
        Y_freq[ff,aa] = np.dot(np.conj(steering_vect), steering_vect_target) / mic_num
        Y_freq_mvdr[ff,aa] = np.dot(np.conj(w), steering_vect)
```

```python
Y_target = np.abs(Y_target)**2 # convert into power
Y_target_mvdr = np.abs(Y_target_mvdr)**2 # convert into power

Y_freq = np.abs(Y_freq)**2 # convert into power
Y_freq_mvdr = np.abs(Y_freq_mvdr)**2 # convert into power
```

```python
DI_freq = 10 * np.log10(Y_target / np.mean(Y_freq, axis=1)) # target is compared against the mean azimuth for CBF
DI_freq_mvdr = 10 * np.log10(Y_target_mvdr / np.mean(Y_freq_mvdr, axis=1)) # target is compared against the mean azimuth for MVDR
```

```python
plt.figure()
plt.plot(freq_bins,DI_freq,label="CBF")
plt.plot(freq_bins,DI_freq_mvdr,label="MVDR")
plt.title("Directivity Index (Main Source)")
plt.xlabel("Frequency (Hz)")
plt.xlim(0,400)
plt.ylabel("DI (dB)")
plt.legend()
```

## 13. Beam Pattern

```python
freq_list = [10,100,200,300] # Hz (select frequency)
```

```python
azi_scan_deg = np.arange(0, 360, 1)
ele_scan_deg = np.zeros_like(azi_scan_deg)
azi = np.deg2rad(azi_scan_deg) #rad
ele = np.deg2rad(ele_scan_deg) #rad
```

```python
Y_freq = np.zeros((len(freq_list),len(azi)), dtype=complex)
Y_freq_mvdr = np.zeros((len(freq_list),len(azi)), dtype=complex)

#ENDFIRE
u_beam = np.array([
    np.cos(0) * np.sin(0),
    np.cos(0) * np.cos(0),
    np.sin(0)
    ])

# 90 DEGREES
# u_beam = np.array([
#     np.cos(0) * np.sin(np.deg2rad(90)),
#     np.cos(0) * np.cos(np.deg2rad(90)),
#     np.sin(0)
#     ])

tau_beam = -np.dot(mic_pos,u_beam) / c
```

```python
for ff, freq in enumerate(freq_list):
    
    steering_vect_target = np.exp(-1j * 2 * np.pi* freq * tau_beam)
    freq_index = np.argmin(np.abs(freq_bins-freq))
    R = np.sinc(2 * freq * d / c)
    loading = 0.1 * np.trace(R).real / mic_num
    R_loaded = R + loading * np.eye(mic_num)
    Rinv = np.linalg.inv(R_loaded)
    w = (Rinv @ steering_vect_target)/(steering_vect_target.conj().T @ Rinv @ steering_vect_target)                

    for aa in range(len(azi)):
        
        u = np.array([
            np.cos(ele[aa]) * np.sin(azi[aa]),
            np.cos(ele[aa]) * np.cos(azi[aa]),
            np.sin(ele[aa])
            ])
    
        tau = -np.dot(mic_pos,u) / c
    
        steering_vect = np.exp(-1j * 2 * np.pi* freq * tau)
        Y_freq[ff,aa] = np.dot(np.conj(steering_vect), steering_vect_target) / mic_num
        Y_freq_mvdr[ff,aa] = np.dot(np.conj(w), steering_vect)
```

```python
Y_freq = np.abs(Y_freq)**2
Y_freq = Y_freq / np.max(Y_freq, axis=1, keepdims=True) # normalise
Y_azi_dB = 10*np.log10(Y_freq + 1e-12)

Y_freq_mvdr = np.abs(Y_freq_mvdr)**2
Y_freq_mvdr = Y_freq_mvdr / np.max(Y_freq_mvdr, axis=1, keepdims=True) # normalise
Y_azi_mvdr_dB = 10*np.log10(Y_freq_mvdr + 1e-12)
```

```python
fig = plt.figure(figsize=(25, 5))
specgrid = GridSpec(nrows=1,ncols=len(freq_list),figure=fig)
#------------------------------------------------------------------------------
# BEAM PATTERN COMPARISON PLOT
for i in range(len(freq_list)):
    ax = fig.add_subplot(specgrid[i], projection='polar')
    ax.plot(azi,Y_azi_dB[i,:],label="CBF")
    ax.plot(azi,Y_azi_mvdr_dB[i,:],label="MVDR")
    ax.set_theta_zero_location("N")
    ax.set_theta_direction(-1)
    ax.set_title(f"Azimuth Scan for Frequency {freq_list[i]} Hz")
    ax.set_xlabel("Azimuth (degrees)")
    ax.set_ylabel("Magnitude (dB)")
    ax.legend()
```
